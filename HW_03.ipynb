```python
def sigmoid(z):
  z = np.array(z)
  return 1/(1+np.exp(-z))
```

```python
def train_logistic_regression(x,y,lr = 0.01, epochs = 10000):

  x = np.array(x, dtype = np.float64)
  y = np.array(y, dtype = np.float64)

  m, n = x.shape

  weights = np.zeros(n)


  for epoch in range(epochs):
    z = np.dot(x, weights)
    preds = sigmoid(z)
    gradient = (1/m)*np.dot(x.T, (preds-y))
    weights -= lr*gradient

    if epoch % 1000 == 0:
      loss =  - (1/m) * (np.dot(y, np.log(preds + 1e-9)) + np.dot(1 - y, np.log(1 - preds + 1e-9)))
      print(f'Epoch {epoch}: Loss: {loss: .4f}')

  return weights


def predict(x, weights, threshold = 0.5):

  x = np.array(x, dtype = np.float64)
  return(sigmoid(np.dot(x, weights)) >= threshold).astype(int)
```

```python
# Use the encoded and scaled X_train / X_test!
weights = train_logistic_regression(X_train, y_train)
y_pred = predict(X_test, weights)

from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy}')
```
