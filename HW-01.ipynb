```python

import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from sklearn import preprocessing
from sklearn.preprocessing import MinMaxScaler
# Option 1, load from sample_data if available
boston_df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv', sep = ",")
boston_df.head()
print("Print first 10 rows of the data frame")
print(boston_df.head(10))
boston_df.info()
print("statistical summary of the data frame")
print(boston_df.describe())
print("Count of missing values in each column:")
print(boston_df.isnull().sum())
mean_values = boston_df['age'].mean() #calculating mean of  [desired column]
print("Mean value of age column:", mean_values)
boston_df['age'] = boston_df['age'].fillna(mean_values)#replacing null values in the column with mean values
print("Count of missing values in each column after filling:")
```

```python
import pandas as pd 
from sklearn.preprocessing import MinMaxScaler , StandardScaler, Binarizer
#Min-Max scaling is a technique used to scale the data to a fixed range, usually 0 to 1. It is done by subtracting the minimum value of the feature and then dividing by the range of the feature
min_max_scaler = MinMaxScaler() #creating an object of MinMaxScaler class
boston_df_min_max_scaled = pd.DataFrame(min_max_scaler.fit_transform(boston_df), columns=boston_df.columns) #fitting the data to the scaler and transforming it to a dataframe
print("Data frame after Min-Max scaling:")
print(boston_df_min_max_scaled.head()) #displaying the first 5 rows of the data frame after Min-Max scaling min_max takes the minimum value of each column and subtracts it from each value in that column, then divides by the range of that column (max-min) to scale the values between 0 and 1
boston_df_min_max_scaled.describe() #to get the statistical summary of the data frame after Min-Max scaling use df.describe()
print("Statistical summary of the data frame after Min-Max scaling:")
print(boston_df_min_max_scaled.describe())
```

```python
#standardscaling is a technique used to scale the data to have a mean of 0 and a standard deviation of 1. It is done by subtracting the mean of the feature and then dividing by the standard deviation of the feature. The formula for standard scaling is:
standard_scaler = StandardScaler() #creating an object of StandardScaler class
boston_df_standard_scaled = pd.DataFrame(standard_scaler.fit_transform(boston_df), columns=boston_df.columns) #fitting the data to the scaler and transforming it to a dataframe
print("Data frame after Standard scaling:")
print(boston_df_standard_scaled.head()) #displaying the first 5 rows of the data frame after Standard scaling
boston_df_standard_scaled.describe() #to get the statistical summary of the data frame after Standard scaling use df.describe()
print("Statistical summary of the data frame after Standard scaling:")
print(boston_df_standard_scaled.describe())
```
